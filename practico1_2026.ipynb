## Parte 2: *Condicionales*


Las instrucciones condicionales permiten que un programa tome decisiones
en función del valor de una **expresión lógica**. Dependiendo de si la
condición es verdadera o falsa, se ejecutará un bloque de instrucciones
u otro.

En Python, las condiciones se construyen a partir de expresiones que
devuelven valores booleanos:

```
True
False
```

Estas expresiones suelen involucrar:

- **operadores relacionales:** `<`, `>`, `<=`, `>=`, `==`, `!=`
- **operadores lógicos:** `and`, `or`, `not`


La forma más simple de una instrucción condicional en Python es:

```python
if condicion:
    bloque_de_instrucciones
```

Primero se evalúa la condición.

- Si la condición es `True`, se ejecuta el bloque indentado.
- Si la condición es `False`, el bloque no se ejecuta.

Por ejemplo

```python
x = 5

if x > 0:
    print("El número es positivo")
```

Muchas veces queremos ejecutar una acción alternativa cuando la condición
no se cumple. Para ello se utiliza `else`.

```python
if condicion:
    bloque_1
else:
    bloque_2
```
Por ejemplo 

```python
x = -3

if x > 0:
    print("El número es positivo")
else:
    print("El número no es positivo")
```

Cuando hay varias condiciones posibles se puede utilizar `elif`.

```python
if condicion1:
    bloque1
elif condicion2:
    bloque2
else:
    bloque3
```

Veamos

```python
x = 0

if x > 0:
    print("Positivo")
elif x < 0:
    print("Negativo")
else:
    print("Cero")
```

**Buenas prácticas al usar condicionales**

El uso adecuado de condicionales ayuda a que el código sea claro,
legible y fácil de mantener.

- Escribir condiciones claras. Evitar condiciones innecesariamente complicadas.
- Si una variable ya es booleana, no es necesario compararla con `True`.
Por ejemplo esto seria correcto

```python
if es_valido:
    print("Dato válido")
```

y esto incorrecto

```python
if es_valido == True:
    print("Dato válido")
```

- Usar operadores lógicos con claridad.


Las instrucciones condicionales son una herramienta fundamental para
controlar el flujo de ejecución de un programa. Permiten construir
algoritmos que **toman decisiones en función de los datos de entrada
o de resultados intermedios**.

## PARTE 3: *Ciclos*

Los ciclos permiten repetir un conjunto de instrucciones varias veces.  
Son una herramienta fundamental para implementar algoritmos que requieren
realizar la misma operación sobre distintos datos o durante varias iteraciones.

En Python existen dos tipos principales de ciclos:

- `for`
- `while`

Su diferencia conceptual radica en que el ciclo `for` se utiliza para recorrer objetos iterables. Un iterable, lo veremos dentro de unas clases, es un objeto que contiene una secuencia de elementos que pueden recorrerse uno a uno. El ciclo `while`, en cambio, no trabaja sobre iterables.  
En lugar de recorrer una secuencia, ejecuta instrucciones mientras una condición logica sea verdadera.

La forma general es:

```python
while condicion:
    bloque_de_instrucciones
```

Recordemos el ejemplo de fibonacci.

```python
a, b = 0, 1
n = 10

while n > 0:
    print(a)
    a, b = b, a + b
    n = n - 1
```

En este ejemplo:

- $n$ controla cuántas veces se ejecuta el ciclo

- mientras $n > 0$ el ciclo continúa

- en cada iteración se actualizan los valores de $a$ y $b$

- $n$ disminuye en cada paso hasta que la condición deja de cumplirse

Buenas prácticas al usar while:

- Usar condiciones claras

- Modificar correctamente las variables de control

- Evitar ciclos infinitos accidentales

Los ciclos permiten construir algoritmos que realizan procesos repetitivos,
como recorrer datos, acumular valores o aplicar cálculos varias veces.

El uso adecuado de ciclos es fundamental para resolver problemas
computacionales de manera clara y eficiente.

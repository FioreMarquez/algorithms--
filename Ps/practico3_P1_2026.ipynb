## PARTE 1: *Ciclos*

Los ciclos permiten **repetir un conjunto de instrucciones varias veces**.  
Son una herramienta fundamental para implementar algoritmos que requieren
realizar la misma operación sobre distintos datos o durante varias iteraciones.

En Python existen dos tipos principales de ciclos:

- `for` → cuando sabemos cuántas veces queremos repetir una operación.
- `while` → cuando repetimos mientras una condición sea verdadera.

En esta sección nos enfocaremos principalmente en el ciclo `for`. El ciclo `for` se utiliza para **recorrer una secuencia de valores**.

La forma general es:

```python
for variable in iterable:
    bloque_de_instrucciones
```

En cada iteración:

1. la variable toma un valor de la secuencia
2. se ejecuta el bloque de instrucciones


```python
for i in range(5):
    print(i)
```

Salida:

```
0
1
2
3
4
```

En este caso la variable `i` va tomando los valores generados por `range(5)`.

La función `range` genera secuencias de números,  es un iterable especial que:

- representa una secuencia de números
- no guarda todos los valores en memoria
- los genera cuando los necesitás

### Forma básica

```python
range(start, stop, step)
```

- `start`: Es opcional, entero que especifica el numero donde comienza la secuencia. Default es 0
- `stop`: Es requerido, entero que especifica el numero donde finaliza la secuencia (sin incluirlo)
- `step`: Es opcional, entero que especifica el paso de incrementacion.	Ddefault 1.

**Buenas prácticas al usar ciclos**
- Usar nombres de variables claros
- Mantener el cuerpo del ciclo simple
- Usar `range` cuando se conoce el número de iteraciones


Los ciclos permiten construir algoritmos que realizan **procesos repetitivos**,
como recorrer datos, acumular valores o aplicar cálculos varias veces.

El uso adecuado de ciclos es fundamental para resolver problemas
computacionales de manera clara y eficiente.

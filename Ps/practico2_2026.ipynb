## *Funciones*

En programación, una función es un bloque de código que realiza una tarea específica. Permiten organizar los programas, evitar repetir código y hacer que los programas sean más fáciles de leer y mantener.

En términos matemáticos, una función puede pensarse como una transformación:

$$
f(x) \rightarrow y
$$


Muchas veces en un programa necesitamos repetir un mismo cálculo varias veces. Como ha ocurrido en el practico 1 con el ingreso de datos a traves del usuario.

Por ejemplo:

```python
area1 = 3.1416 * r1**2
area2 = 3.1416 * r2**2
area3 = 3.1416 * r3**2
```

En lugar de repetir el cálculo, podemos definir una función:

```python
def area_circulo(r):
    return 3.1416 * r**2
```

Luego podemos reutilizarla:

```python
area1 = area_circulo(r1)
area2 = area_circulo(r2)
area3 = area_circulo(r3)
```

La forma general para definir una función en Python es:

```python
def nombre_funcion(parametros):
    instrucciones
    return resultado
```

Componentes:

- `def` → palabra reservada que indica la definición de una función
- `nombre_funcion` → nombre elegido para la función
- `parametros` → variables que recibe la función
- `cuerpo` → instrucciones que realiza la función
- `return` → valor que devuelve la función

Un ejemplo:

```python
def cuadrado(x):
    return x * x
```
y luego lo usamos como

```python
cuadrado(5)
```
La función se ejecuta cuando se llama. En el caso anterior:

```python
resultado = cuadrado(4)
print(resultado)
```

Proceso de ejecución:

1. Se evalúan los argumentos
2. Se ejecuta el cuerpo de la función
3. Se devuelve el resultado

**Parámetros y argumentos**

- Parámetro formal. Es la variable que aparece en la definición de la función.

```python
def cuadrado(x):
```

En este caso, `x` es el parámetro formal. Pueden ser varios.

- Parámetros con valores por defecto. Esto significa que si al llamar a la función no se proporciona ese argumento, la función utilizará el valor definido por defecto.

Por ejemplo:

```python
def saludar(nombre="mundo"):
    print("Hola", nombre)
```

Podemos llamar a la función sin argumentos:

```python
saludar()
```

y obtendremos

```
Hola mundo
```

Pero tambien podemos pasar un argumento distinto:

```python
saludar("Ana")
```

lo que produce

```
Hola Ana
```

Esto ocurre porque `"mundo"` es el valor por defecto del parámetro `nombre`.

Una regla importante es que los parámetros con valores por defecto deben ir al final de la lista de parámetros.

Por ejemplo, esto es correcto:

```python
def ejemplo(a, b=0):
    return a + b
```

pero esto no es válido y les dara error

```python
def ejemplo(a=0, b):
    return a + b
```


**Funciones que no devuelven valor**

No todas las funciones necesitan devolver un valor. Como puede ser

```python
def saludar(nombre):
    print("Hola", nombre)
```
Entonces

```python
saludar("Ana")
```
da como salida 

```
Hola Ana
```

**Variables locales**

Las variables definidas dentro de una función existen solo dentro de la función.Entonces en

```python
def ejemplo():
    x = 5
    print(x)
```

la variable `x` no existe fuera de la función.


Las funciones son una herramienta fundamental para estructurar programas.  
Permiten encapsular cálculos, reutilizar código y construir programas más claros y organizados.

## *Mas sobre Funciones*

**Anotaciones de tipo (Type Hints)**

En Python es posible indicar el tipo esperado de los parámetros y del valor de retorno de una función.  Esto se hace mediante anotaciones de tipo (*type hints*). No son obligatorias y Python no las verifica estrictamente durante la ejecución.

Su objetivo principal es:

- mejorar la legibilidad del código
- ayudar a herramientas de análisis estático 
- facilitar el autocompletado y detección de errores en editores de código

La sintaxis básica:

Las anotaciones de tipo se agregan después del nombre del parámetro utilizando `:`  
y el tipo de retorno se indica con `->`.

```python
def cuadrado(x: float) -> float:
    return x * x
```

En este ejemplo:

- `x: float` indica que el parámetro `x` se espera que sea un número real.
- `-> float` indica que la función devuelve un número real.

**Importante**

Las anotaciones no obligan a Python a respetar los tipos.

Por ejemplo:

```python
def cuadrado(x: float) -> float:
    return x * x

print(cuadrado("hola"))
```

Este código no produce un error de tipos automáticamente en Python.  
Las anotaciones son principalmente informativas.

**Signatura de una función**

Describe cómo debe ser utilizada la función.  
Indica:

- el nombre de la función
- los parámetros que recibe
- los tipos esperados de esos parámetros (si se indican)
- el tipo del valor que devuelve

Por ejemplo, si definimos:

```python
def cuadrado(x: float) -> float:
    return x * x
```

la signatura de la función es:

```python
cuadrado(x: float) -> float
```

La signatura funciona como una especie de contrato de uso de la función, ya que describe cómo debe ser llamada y qué tipo de resultado produce.

**Documentación de funciones (Docstrings)**

Las funciones pueden incluir una cadena de documentación, llamada *docstring*,
que explica qué hace la función, cuáles son sus parámetros y qué devuelve.

La docstring se escribe como una cadena entre comillas triples `"""` al comienzo de la función.

Ejemplo de type hints y de documentacion 

```python
def area_circulo(r: float) -> float:
    """
    Calcula el área de un círculo.

    Parámetros
    ----------
    r : float
        Radio del círculo.

    Retorna
    -------
    float
        Área del círculo.
    """
    import math
    return math.pi * r**2
```


La documentación de una función puede consultarse usando:

```python
help(area_circulo)
```

o

```python
print(area_circulo.__doc__)
```


Las anotaciones de tipo y las docstrings ayudan a escribir código más claro y mantenible.  
Aunque Python no obliga a respetar los tipos, estas herramientas facilitan la comprensión del programa
y son ampliamente utilizadas en proyectos profesionales.

````python

def doble(x: float)-> float:
    """
    Esta funcion calcula el doble del numero x
    lo hace asi
    x*2
    """
    return x*2
````

## *Mas sobre Funciones*

**Verificación de tipos con `isinstance`**

Aunque Python permite usar anotaciones de tipo (type hints), estas no se verifican automáticamente durante la ejecución del programa.  
Si queremos asegurarnos de que los argumentos de una función tienen el tipo esperado, podemos realizar verificaciones explícitas dentro de la función.

Para ello se utiliza la función incorporada `isinstance()`. La función `isinstance(objeto, tipo)` devuelve:

- `True` si el objeto pertenece al tipo indicado
- `False` en caso contrario

Por ejemplo

```python
x = 5
print(isinstance(x, int))
print(isinstance(x, float))
```

Nos va a dar 

```
True
False
```

Un ejemplo mas

```python
x = 3.14
y = "hola"
z = True

print(isinstance(x, float))
print(isinstance(y, str))
print(isinstance(z, bool))
```

Podemos usar `isinstance` dentro de una función para verificar que los parámetros tengan el tipo esperado.

```python
def cuadrado(x):
    if not isinstance(x, (int, float)):
        print("Error: el argumento debe ser un número")
        return

    return x * x
```

Lo usamos como  antes

```python
print(cuadrado(5))
print(cuadrado(3.2))
```

Pero si intentamos 

```python
cuadrado("hola")
```

el programa mostrará:

```
Error: el argumento debe ser un número
```

Como acabamos de ver en el ejemplo, es posible verificar varios tipos al mismo tiempo utilizando una tupla de tipos. En unas clases mas adelante veremos que es una tupla, por ahora piensenlo como un vector.

```python
isinstance(x, (int, float))
```

Esto significa:

```
x es int o x es float
```

Ejemplo:

```python
def es_numero(x):
    return isinstance(x, (int, float))
```


Por supuesto podemos combinar type hints con verificación explícita.

```python
def promedio(a: float, b: float) -> float:
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        print("Los argumentos deben ser números")

    return (a + b) / 2
```

Las anotaciones ayudan a documentar el código, mientras que `isinstance`
permite realizar verificaciones reales durante la ejecución.



En Python es común confiar en el tipado dinámico del lenguaje.  
Sin embargo, en funciones críticas o bibliotecas es útil usar `isinstance`
para verificar que los argumentos recibidos sean del tipo esperado.

Esto ayuda a:

- detectar errores temprano
- producir mensajes de error más claros
- escribir funciones más robustas


**Verificación de condiciones con `assert`**

Además de verificar tipos con `isinstance`, Python permite verificar condiciones que deben cumplirse al entrar a una función utilizando la instrucción `assert`.

Un `assert` se utiliza para expresar suposiciones que el programa espera que sean verdaderas.  
Si la condición es falsa, Python detiene la ejecución y muestra un mensaje de error.

La forma general es:

```python
assert condicion, "mensaje de error"
```

Si `condicion` es verdadera, el programa continúa normalmente.  
Si es falsa, Python produce un **AssertionError** junto con el mensaje indicado.

Por ejemplo

```python
x = 5
assert x > 0, "x debe ser positivo"
```

Si la condición no se cumple:

```python
x = -3
assert x > 0, "x debe ser positivo"
```

Python mostrará:

```
AssertionError: x debe ser positivo
```



Los `assert` son útiles para expresar precondiciones de una función. Es decir una condición que los argumentos deben cumplir antes de ejecutar la función.

Por ejemplo, si una función espera números positivos, podemos verificarlo al comienzo.

```python
def raiz(x):
    assert x >= 0, "x debe ser mayor o igual a cero"
    import math
    return math.sqrt(x)
```

También es posible escribir las verificaciones de tipo utilizando `isinstance`.

Por ejemplo:

```python
assert isinstance(d, int), "el día debe ser entero"
```


Los `assert` se utilizan principalmente para expresar precondiciones y suposiciones del programa.  
Son muy útiles para detectar errores durante el desarrollo y documentar qué condiciones deben cumplir los argumentos de una función.

## *Funciones lambda en Python*

Las funciones lambda son funciones anónimas (es decir, sin nombre) que se definen en una sola línea.

Son útiles cuando necesitamos una función simple y corta por única vez, sin necesidad de definirla con `def`.

**Sintaxis**

```python
lambda argumentos: expresión
```

* argumentos: los parámetros que recibe la función

* expresión: lo que la función devuelve (no hace falta usar return)

Ejemplo:

```python
cuadrado = lambda x: x**2

print(cuadrado(3))  # 9
```

es equivalente a 
```python
def cuadrado(x):
    return x**2
```

Una lambda es simplemente una forma corta de escribir funciones pequeñas.


## *Composicion de funciones*

En matemática, la composición de funciones consiste en aplicar una función sobre el resultado de otra.

Si tenemos dos funciones $f$ y $g$, la composición se escribe:

$$
(f \circ g)(x) = f(g(x))
$$

entonces primero se aplica $g$  yl uego se aplica $f$ sobre ese resultado  

**Ejemplo matemático**

Sean:

$$
g(x) = x + 1
$$

$$
f(x) = 2x
$$

Entonces:

$$
(f \circ g)(x) = f(g(x)) = 2(x + 1)
$$

**Ejemplo en Python**

```python
def g(x):
    return x + 1

def f(x):
    return 2 * x

resultado = f(g(3))  # primero g(3)=4, luego f(4)=8
print(resultado)     # 8
```

### *Relación con lambda*

La composición suele implementarse con lambda porque:

- queremos devolver una función

- esa función suele ser simple

El número combinatorio $C(n,k)$, también escrito como “n sobre k”, indica de cuántas maneras se pueden elegir $k$ elementos de un conjunto de $n$ elementos, sin importar el orden.

Se define por la fórmula:

$$
C(n,k) = \frac{n!}{k!(n-k)!}
$$

En este ejercicio se busca que reutilices la función `factorial` del ejercicio anterior y luego construyas una nueva función con `lambda`.


1. Escribir una función llamada `combinaciones(n, k)` que:
   - reciba dos argumentos: `n` y `k`
   - devuelva el valor de:

$$
C(n,k) = \frac{n!}{k!(n-k)!}
$$

2. La función debe verificar que:
   - `n` sea un entero
   - `k` sea un entero
   - `n >= 0`
   - `k >= 0`
   - `k <= n`

3. Si alguna de estas condiciones no se cumple, debe producir un error con `assert`.

4. Luego, definir una función lambda llamada `combinaciones_lambda` que:
   - reciba dos argumentos, `n` y `k`
   - calcule exactamente el mismo valor que `combinaciones(n, k)`
   - use la función `factorial` ya implementada

5. Probar ambas funciones con distintos ejemplos.

#### Ejemplo de comportamiento esperado

```python
combinaciones(5, 2)   # 10
combinaciones(6, 0)   # 1
combinaciones(6, 6)   # 1

combinaciones_lambda(5, 3)   # 10

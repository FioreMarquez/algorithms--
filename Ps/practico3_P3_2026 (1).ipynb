---

En Python, `map()` es una función que permite aplicar otra función a cada elemento de un iterable (lista, tupla, etc.), devolviendo un nuevo iterable con los resultados. Sintaxis basica:

```python
map(funcion, iterable)
```
- funcion: la función que querés aplicar
- iterable: lista, tupla, etc.

En Python 3, map devuelve un iterador, por lo que muchas veces hay que convertirlo a lista `list(map(...))`

Ejemplo con lambda:

````python
numeros = [1, 2, 3, 4]

resultado = list(map(lambda x: x*2, numeros))

print(resultado)  
````

Ejemplo con cadenas

````python
palabras = ["hola", "mundo"]

mayusculas = list(map(str.upper, palabras))

print(mayusculas)  
````

### Ejercicio 9:

Definir una función llamada `norma(v, p=2)` que:

- Reciba como parámetro un vector representado mediante una lista de números.
- Retorne la norma $p$ del vector.

**Definición matemática**

La norma $p$ de un vector $v = (v_1, v_2, \dots, v_n)$ está dada por:

- Si $1 \leq p < \infty$:

$$
\|v\|_p = \left( \sum_{i=1}^{n} |v_i|^p \right)^{1/p}
$$

- Si $p \to \infty$:

$$
\|v\|_\infty = \max_i |v_i|
$$


- Validar que:
  - El argumento recibido sea una lista.
  - Todos sus elementos sean valores numéricos (`int` o `float`).
- En caso de lista vacía, la función debe retornar `0`.
- El parámetro `p` debe ser positivo o igual a `float('inf')`.
- No modificar la lista original.

Ejemplo de uso esperado

```python
norma([3, 4])                     # 5.0  (norma euclídea)
norma([3, 4], p=1)               # 7    (norma 1)
norma([3, 4], p=float('inf'))    # 4
```

Dado $v = (v_1, v_2, \dots, v_n)$ y $w = (w_1, w_2, \dots, w_n)$, el producto escalar está definido como:

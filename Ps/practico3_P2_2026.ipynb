## PARTE 2: *Cadenas*

Las **cadenas de caracteres** (str) son uno de los tipos de datos más utilizados en Python.
Una cadena es una secuencia ordenada de caracteres (texto). 

- Creación de cadenas:
```python
a = "Hola"
b = 'Python'
c = """Esto es
una cadena multilínea"""
```
Se pueden usar comillas simples, dobles o triples.

- Las cadenas son indexables, índices negativos acceden desde el final


````python
texto = "Python"

print(texto[0])   # 'P'
print(texto[1])   # 'y'
print(texto[-1])  # 'n'
````

- Rebanado (slicing), se puede seccionar parte de la cadena de la siguiente forma

```python
cadena[inicio:fin:paso]
```
entonces en el ejemplo


````python
texto = "Python"

print(texto[0:3])   # 'Pyt'
print(texto[2:])    # 'thon'
print(texto[:4])    # 'Pyth'
print(texto[::2])   # 'Pto'
````

- Las cadenas son iterables, por lo que se pueden recorrer con `for`:


````python
texto = "Hola"

for letra in texto:
    print(letra)
````

````python
for i in range(len(texto)):
    print(i, texto[i])
````

- Longitud de una cadena:

````python
len(texto) 
````

- Operaciones básicas: 


````python
# Concatenación
"Hola" + " " + "mundo"
````

````python
# Repetición
"ha" * 3   # 'hahaha'
````

````python
# Pertenencia
print("a" in "casa")   # True
print("z" not in "casa")  # True
````

- Métodos más importantes de cadenas:
    - Conversión de mayúsculas/minúsculas:


````python
texto = "Hola Mundo"

print(texto.lower())   # 'hola mundo'
print(texto.upper())   # 'HOLA MUNDO'
print(texto.capitalize())  # 'Hola mundo'
print(texto.title())   # 'Hola Mundo'
````

- Métodos más importantes de cadenas:
    - Eliminación de espacios:

````python
texto = "  hola  "

print(texto.strip())    # 'hola'
print(texto.lstrip())   # 'hola  '
print(texto.rstrip())   # '  hola'
````

- Métodos más importantes de cadenas:
    - Búsqueda:

````python
texto = "Hola mundo"

print(texto.find("mundo"))   # posición inicial o -1 si no está
print(texto.index("Hola"))   # como find pero lanza error si no está
````

- Métodos más importantes de cadenas:
    - Conteo:

````python
texto = "banana"

texto.count("a")   # 3
````

- Métodos más importantes de cadenas:
    - Reemplazo:

````python
texto = "Hola mundo"

texto.replace("mundo", "Python")  # 'Hola Python'
````

- Métodos más importantes de cadenas:
    - División (split):



````python
texto = "uno,dos,tres"

texto.split(",")   # ['uno', 'dos', 'tres']
````

 - Métodos más importantes de cadenas:
    - verificaciones: 

````python
print("abc".isalpha())    # Devuelve True si todos los caracteres son letras, no permite numeros ni espacios
print("123".isdigit())    # Devuelve True si todos los caracteres son dígitos
print("abc123".isalnum()) # Devuelve True si todos los caracteres son letras o números
print("hola".islower())   # Devuelve True si todas las letras están en minúscula
print("HOLA".isupper())   # Devuelve True si todas las letras están en mayúscula
print("Hola".istitle())   # Devuelve True si el string está en formato título ie cada palabra empieza con una Mayus
````

- Algo **MUY** importante es que las cadenas son inmutables, es decir **no** se pueden modificar directamente.

````python
texto = "hola"

texto[0] = "H"
````

- Caracteres importantes

````python
print("Hola\nMundo")   # salto de línea
print("Hola\tMundo")   # tabulación
print("\"Hola\"")      # comillas
````

- las cadenas se pueden comparar, lo que compara es caracter por caracter usando el orden [Unicode](https://elcodigoascii.com.ar/). Que es? Es un sistema que asigna números a caracteres básicos ya que las computadoras no entienden letras, solo números. Unicode incluye prácticamente todos los caracteres del mundo (idiomas,símbolos,emojis)
    - Igualdad:

````python
print("abc" == "abc" )  # True
print("abc" == "abd")   # False
````

- comparacion:
    - orden:

````python
"abc" < "abd"   # porque 'c' < 'd'
````

- comparacion:
    - por longitud:

````python
"abc" < "abc "  # recordemos que los espacios tambien son caracteres
````

**3.** Implementar una funcion que:

Ejemplo de uso 
```python

```

````python
# codigo

def analizar_cadena1(cadena:str):
    caracteres= 0
    digitos = 0
    letras = 0
    for i in cadena:
        if i.isalpha():
            letras = letras + 1
        if i.isdigit():
            digitos = digitos + 1
        if i.isalnum():
            caracteres = caracteres + 1
    return letras, digitos, caracteres
````

````python
print(analizar_cadena1("hola 123"))
````

````python
def analizar_cadena3(cadena:str):
    espacios = 0
    for i in cadena:
        if i == " ":
            espacios = espacios + 1
    return espacios
````

````python
print(analizar_cadena3(" Hola mundo 123 "))
````

Implementar una función que invierta una cadena sin usar slicing [::-1].

````python
# codigo
def invertir_cadena(cadena:str):
    cadena_invertida = ""
    for i in cadena:
        cadena_invertida = i + cadena_invertida
    return cadena_invertida
````
